```python
import os
import matplotlib.pyplot as plt
from PIL import Image

def mostrar_muestras_por_clase(base_path, clases, n=5):
    fig, axs = plt.subplots(len(clases), n, figsize=(n * 3, len(clases) * 3))
    for i, cls in enumerate(clases):
        cls_path = os.path.join(base_path, cls)
        imgs = [f for f in os.listdir(cls_path) if f.endswith('.png')][:n]
        for j in range(n):
            img = Image.open(os.path.join(cls_path, imgs[j]))
            axs[i, j].imshow(img, cmap='gray')
            axs[i, j].axis('off')
            axs[i, j].set_title(imgs[j], fontsize=7)
        axs[i, 0].set_ylabel(cls, fontsize=10, rotation=0, labelpad=50, va='center')
    
    fig.suptitle("Visualización de muestras por clase (carpeta 'train')", fontsize=14)
    plt.tight_layout(rect=[0, 0, 1, 0.95])  # deja espacio para el título
    plt.show()


base_path = "Data/train"
clases = [
    'adenocarcinoma_left.lower.lobe_T2_N0_M0_Ib',
    'large.cell.carcinoma_left.hilum_T2_N2_M0_IIIa',
    'normal',
    'squamous.cell.carcinoma_left.hilum_T1_N2_M0_IIIa'
]

mostrar_muestras_por_clase(base_path, clases)
```
