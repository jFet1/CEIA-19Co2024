## Preprocesamiento

```python
input_path = 'texto_las-cronicas-de-narnia-el-leon-la-bruja-y-el-ropero.txt'
output_path = 'texto_las-cronicas-de-narnia-el-leon-la-bruja-y-el-ropero_limpio.txt'

omit = False

with open(input_path, 'r', encoding='utf-8') as file, open(output_path, 'w', encoding='utf-8') as output:
    for line in file:
        
        if "Texto de dominio público." in line:
            omit = True
            continue
        if omit and "Padrino que te quiere mucho" in line:
            omit = False
            continue
        if omit:
            continue
        if (
            not line.strip().startswith(tuple(str(i) for i in range(10))) and
            "www.infotematica.com.ar" not in line and
            "Clive S. Lewis" not in line and
            "C. S. LEWIS" not in line
        ):
            output.write(line)

print(f"Archivo preprocesado guardado en: {output_path}")
```

```python
import re

input_path = 'texto_las-cronicas-de-narnia-el-leon-la-bruja-y-el-ropero_limpio.txt'
output_path = 'texto_las-cronicas-de-narnia-el-leon-la-bruja-y-el-ropero_oraciones.txt'

with open(input_path, 'r', encoding='utf-8') as file, open(output_path, 'w', encoding='utf-8') as output:
    text = file.read().replace('\n', ' ')
    sentences = re.split(r'(?<=\.)\s+(?=[A-ZÁÉÍÓÚÑ])', text)
    for sentence in sentences:
        output.write(sentence.strip() + '\n')

print(f"Archivo preprocesado y teniendo una linea por oración guardado en: {output_path}")
```
