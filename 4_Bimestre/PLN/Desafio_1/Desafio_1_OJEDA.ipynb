### Vectorización de texto y modelo de clasificación Naïve Bayes con el dataset 20 newsgroups

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.naive_bayes import MultinomialNB, ComplementNB
from sklearn.metrics import f1_score
import pandas as pd

# 20newsgroups por ser un dataset clásico de NLP ya viene incluido y formateado
# en sklearn
from sklearn.datasets import fetch_20newsgroups
import numpy as np

doc1 = 1010
doc2 = 2020
doc3 = 3030
doc4 = 4040
doc5 = 5050
docs = [doc1, doc2, doc3, doc4, doc5]
```

## Carga de datos

```python
# cargamos los datos (ya separados de forma predeterminada en train y test)
newsgroups_train = fetch_20newsgroups(subset='train', remove=('headers', 'footers', 'quotes'))
newsgroups_test = fetch_20newsgroups(subset='test', remove=('headers', 'footers', 'quotes'))
```

## Similaridad de documentos

```python
tfidfvect = TfidfVectorizer()
X_train = tfidfvect.fit_transform(newsgroups_train.data)
y_train = newsgroups_train.target
cossim = cosine_similarity(X_train[docs], X_train)
```

```python
# Buscamos los documentos más similares a los seleccionados
for i, doc_idx in enumerate(docs):
    similares = np.argsort(cossim[i])[::-1][1:6]
    print(f"Documento base: {doc_idx} - Clase: {newsgroups_train.target_names[y_train[doc_idx]]}")
    for sim_idx in similares:
        print(f"   Similar: {sim_idx} - Clase: {y_train[sim_idx]} - Target: {newsgroups_train.target_names[y_train[sim_idx]]}")
```

### 2. Modelos de clasificación Naïve Bayes

```python
def comparar_modelos_nb(train_data, test_data, y_train, y_test):
    resultados = []
    
    # Distintas configuraciones del vectorizador
    vectorizers = [
        ("default", TfidfVectorizer()),
        ("stopwords+bigrams", TfidfVectorizer(
                                                stop_words='english',
                                                ngram_range=(1, 2)
                                                )),
        ("filtered+sublinear", TfidfVectorizer(
                                                stop_words='english',
                                                max_df=0.75, min_df=5,
                                                sublinear_tf=True
                                                )),
    ]
    
    # Distintas configuraciones de modelos
    modelos = [
        ("MultinomialNB α=1.0", MultinomialNB(alpha=1.0)),
        ("MultinomialNB α=0.5", MultinomialNB(alpha=0.5)),
        ("ComplementNB α=1.0", ComplementNB(alpha=1.0)),
        ("ComplementNB α=0.5", ComplementNB(alpha=0.5)),
    ]
    
    for nombre_vect, vect in vectorizers:
        X_train_vect = vect.fit_transform(train_data)
        X_test_vect = vect.transform(test_data)
        
        for nombre_modelo, modelo in modelos:
            modelo.fit(X_train_vect, y_train)
            y_pred = modelo.predict(X_test_vect)
            f1 = f1_score(y_test, y_pred, average='macro')
            
            resultados.append({
                "Vectorizador": nombre_vect,
                "Modelo": nombre_modelo,
                "F1 macro": f1
            })
    
    return pd.DataFrame(resultados).sort_values(by="F1 macro", ascending=False)


resultados_nb = comparar_modelos_nb(newsgroups_train.data,
                                    newsgroups_test.data,
                                    newsgroups_train.target,
                                    newsgroups_test.target)
resultados_nb
```

## Conclusión 

El mejor desempeño se obtuvo con el modelo ComplementNB con alfa igual a 0.5, usando un vectorizador que elimina stopwords y considera bigramas. Esta combinación alcanzó un F1 macro de 0.7105.

Esto sugiere que ComplementNB se adapta mejor a datos desbalanceados, y que el uso de bigramas junto con la eliminación de palabras vacías mejora la representación de los documentos. En contraste, el modelo MultinomialNB sin ajustes tuvo el peor desempeño, lo que evidencia la importancia de probar distintas configuraciones para optimizar resultados.

```python
# Vectorizador con parámetros ajustados para análisis de términos
vectorizador = TfidfVectorizer(stop_words='english', max_df=0.75, min_df=5, sublinear_tf=True)
X = vectorizador.fit_transform(newsgroups_train.data)

# Transponer la matriz para pasar de documento-término a término-documento
X_td = X.T  # n_terms x n_documents

# Vocabulario en orden del índice
terms = np.array(vectorizador.get_feature_names_out())

# Se eligen 5 palabras manualmente, relevantes para el dataset
palabras_objetivo = ['god', 'car', 'windows', 'israel', 'space']

# Se calculan las 5 palabras más similares para cada una
for palabra in palabras_objetivo:
    if palabra in terms:
        idx = np.where(terms == palabra)[0][0]
        sims = cosine_similarity(X_td[idx], X_td).flatten()
        top5_idx = sims.argsort()[::-1][1:6]
        similares = terms[top5_idx]
        print(f"\nPalabra: {palabra}\nPalabras más similares: {', '.join(similares)}")
```

## Conclusión

Se seleccionaron manualmente cinco palabras relevantes del corpus: god, car, windows, israel y space. Para cada una, se calcularon las cinco palabras más similares usando la matriz término-documento y la similitud del coseno.

Los resultados muestran agrupamientos semánticos esperables. Por ejemplo, "god" se asocia con términos religiosos como "jesus" y "bible", mientras que "car" se relaciona con "engine", "dealer" y "civic". También se observan asociaciones válidas en contextos tecnológicos ("windows" con "microsoft" y "files"), políticos ("israel" con "arab" y "lebanon") y científicos ("space" con "nasa" y "shuttle").

Esto confirma que la representación vectorial captura adecuadamente el contexto temático de las palabras.
